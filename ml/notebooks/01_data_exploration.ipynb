# 01 — Data Exploration

**Purpose:** Understand the shape and quality of our Bitcoin price data before modelling.

**Outputs:**
- Cleaned price DataFrame saved to `data/btc_prices.csv`
- Exploratory plots

**Next notebook:** `02_model_training.ipynb`

```python
import requests
import pandas as pd
import matplotlib.pyplot as plt

# Fetch 365 days of BTC/USD data from CoinGecko (no API key needed)
url = 'https://api.coingecko.com/api/v3/coins/bitcoin/market_chart'
params = {'vs_currency': 'usd', 'days': '365'}

response = requests.get(url, params=params)
response.raise_for_status()
data = response.json()

# Parse into DataFrame
df = pd.DataFrame(data['prices'], columns=['timestamp_ms', 'price'])
df['date'] = pd.to_datetime(df['timestamp_ms'], unit='ms')
df = df.drop('timestamp_ms', axis=1).set_index('date')

print(f'Shape: {df.shape}')
print(f'Date range: {df.index.min()} → {df.index.max()}')
print(f'Price range: ${df.price.min():,.0f} → ${df.price.max():,.0f}')
df.head()
```

```python
# Check for missing values
print('Missing values:', df.isnull().sum())

# Plot price history
plt.figure(figsize=(14, 5))
plt.plot(df.index, df.price)
plt.title('Bitcoin Price — Last 365 Days')
plt.ylabel('Price (USD)')
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Save cleaned data
import os
os.makedirs('data', exist_ok=True)
df.to_csv('data/btc_prices.csv')
print('Saved to data/btc_prices.csv')
```
